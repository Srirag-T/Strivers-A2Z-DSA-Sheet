Print Name N times using Recursion

```python
class Solution:
    # Recursive function to print name count times
    def printName(self, name, count, N):
        # Base case: if count equals N, stop recursion
        if count == N:
            return

        # Print the name
        print(name)

        # Recursive call with incremented count
        self.printName(name, count + 1, N)

if __name__ == "__main__":
    sol = Solution()
    N = 5
    name = "Ashish"

    sol.printName(name, 0, N)
```

Print 1 to N using Recursion

```python
class Solution:
    def countNum(self,count,N):
        if count == N+1:
            return
        print(count,end=" ")

        self.countNum(count+1,N)

if __name__=="__main__":
    sol = Solution()
    N = 3
    sol.countNum(1,N)

```

Print N to 1 using Recursion

```python
class Solution:
    def countNum(self,N):
        if N <=0:
            return
        print(N,end=" ")

        self.countNum(N-1)

if __name__=="__main__":
    sol = Solution()
    N = 3
    sol.countNum(N)

```
