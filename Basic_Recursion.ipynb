Print Name N times using Recursion

```python
class Solution:
    # Recursive function to print name count times
    def printName(self, name, count, N):
        # Base case: if count equals N, stop recursion
        if count == N:
            return

        # Print the name
        print(name)

        # Recursive call with incremented count
        self.printName(name, count + 1, N)

if __name__ == "__main__":
    sol = Solution()
    N = 5
    name = "Ashish"

    sol.printName(name, 0, N)
```

Print 1 to N using Recursion

```python
class Solution:
    def countNum(self,count,N):
        if count == N+1:
            return
        print(count,end=" ")

        self.countNum(count+1,N)

if __name__=="__main__":
    sol = Solution()
    N = 3
    sol.countNum(1,N)

```

Print N to 1 using Recursion

```python
class Solution:
    def countNum(self,N):
        if N <=0:
            return
        print(N,end=" ")

        self.countNum(N-1)

if __name__=="__main__":
    sol = Solution()
    N = 3
    sol.countNum(N)

```

Sum of first N Natural Numbers

```python
class Solution:
    def SumNatural(self,N):
        if N == 1:
            return 1

        return N+self.SumNatural(N-1)

if __name__=="__main__":
    sol = Solution()
    N = 6
    result = sol.SumNatural(N)
    print(result)

```

Factorial of a given number

```python
class Solution:
    def Factorial(self,N):
        if N == 1:
            return 1

        return N*self.Factorial(N-1)

if __name__=="__main__":
    sol = Solution()
    N = 5
    result = sol.Factorial(N)
    print(result)

```

Reverse a given Array

```python
class Solution:
    def revArray(self, left, right, N):
        if left >= right:
            return N

        temp = N[left]
        N[left] = N[right]
        N[right] = temp

        return self.revArray(left + 1, right - 1, N)
        

if __name__=="__main__":
    sol = Solution()
    N = [1,2,3,4,5]
    result = sol.revArray(0,len(N)-1,N)
    print(result)

```

Check if the given String is Palindrome or not

```python
class Solution:
    def isPalindrome(self, left, right, Str):
        if left >= right:
            return True

        if Str[left] != Str[right]:
            return False

        return self.isPalindrome(left + 1, right - 1, Str)


if __name__ == "__main__":
    sol = Solution()
    Str = "ABCDCBA"
    result = sol.isPalindrome(0, len(Str) - 1, Str)
    print(result)
```
