Count digits in a number

```python
#Brute Force Approach
class Solution:
    def countDigits(self,N):
        counter=0
        while N>0:
            counter+=1
            N = N // 10
        return counter
            

if __name__ == "__main__":
    sol = Solution()
    N = 329823
    print("N:", N)
    digits = sol.countDigits(N)
    print("Number of Digits in N:", digits)
    
```

```python
#Optimal Approach
import math
class Solution:
    def countDigits(self,N):
        cnt = int(math.log10(N)+1)
        return cnt

if __name__ == "__main__":
    sol = Solution()
    N = 329823
    print("N:", N)
    digits = sol.countDigits(N)
    print("Number of Digits in N:", digits)
    
```
