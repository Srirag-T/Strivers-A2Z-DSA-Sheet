Count digits in a number

```python
#Brute Force Approach
class Solution:
    def countDigits(self,N):
        counter=0
        while N>0:
            counter+=1
            N = N // 10
        return counter
            

if __name__ == "__main__":
    sol = Solution()
    N = 329823
    print("N:", N)
    digits = sol.countDigits(N)
    print("Number of Digits in N:", digits)
    
```

```python
#Optimal Approach
import math
class Solution:
    def countDigits(self,N):
        cnt = int(math.log10(N)+1)
        return cnt

if __name__ == "__main__":
    sol = Solution()
    N = 329823
    print("N:", N)
    digits = sol.countDigits(N)
    print("Number of Digits in N:", digits)
    
```

Reverse Digits of A Number

```python
class Solution:
    def revDigits(self,N):
        result = 0
        while N>0:
            temp = N % 10
            result = result*10+temp
            N = N // 10
        return result
            

if __name__ == "__main__":
    sol = Solution()
    N = 329823
    print("N:", N)
    digits = sol.revDigits(N)
    print("Number of Digits in N:", digits)
    
```

Check if a number is Palindrome or Not

```python
class Solution:
    def Palindrome(self,N):
        num = N
        revNum = 0
        while N>0:
            temp = N % 10
            revNum = revNum*10+temp
            N = N // 10
        if(num == revNum):
            print("It's a Palindrome Number")
        else:
            print("It's not a Palindrome Number")
            
            
if __name__ == "__main__":
    sol = Solution()
    N = 4554
    sol.Palindrome(N)
    
```

Check if a number is Armstrong Number or not

```python
class Solution:
    def Armstrong(self,N):
        num = N
        n = N
        result = 0
        counter=0
        while N>0:
            counter+=1
            N = N // 10

        while n>0:
            temp = n % 10
            result = result+(temp ** counter)
            n = n // 10
        if(num == result):
            print("It's a Armstrong Number")
        else:
            print("It's not a Armstrong Number")
            
            
if __name__ == "__main__":
    sol = Solution()
    N = 153
    sol.Armstrong(N)
    
```

Print all Divisors of a given Number

```python
#Brute Force Approach
class Solution:
    def allDivisors(self,N):
        result = []
        for i in range(1,N+1):
            if(N%i==0):
                result.append(i)
        print(result)
        
            
if __name__ == "__main__":
    sol = Solution()
    N = 12
    sol.allDivisors(N)
    
```

```python
#Optimal Approach
import math
class Solution:
    def allDivisors(self,N):
        result = []
        for i in range(1,int(math.isqrt(N))+1):
            if(N%i==0):
                result.append(i)
                if((N//i)!=i):
                    result.append(N//i)
        print(result)
        
            
if __name__ == "__main__":
    sol = Solution()
    N = 12
    sol.allDivisors(N)
    
```

Check if a number is prime or not

```python
#Brute Force Approach
class Solution:
    def numPrime(self,N):
        cnt = 0
        for i in range(1,N+1):
            if(N%i==0):
                cnt+=1

        if(cnt==2):
            print("It's Prime number")
        else:
            print("It's not a prime number")
        
            
if __name__ == "__main__":
    sol = Solution()
    N = 10
    sol.numPrime(N)
    
```

```python
#Optimal Approach
class Solution:
    def numPrime(self,N):
        cnt = 0
        for i in range(1,int (N**0.5)+1):
            if(N%i==0):
                cnt+=1
                if(N//i!=i):
                    cnt+=1

        if(cnt==2):
            print("It's Prime number")
        else:
            print("It's not a prime number")
        
            
if __name__ == "__main__":
    sol = Solution()
    N = 10
    sol.numPrime(N)
    
```

Find GCD of two numbers

```python
#Brute force
import math
class Solution:
    def GCD(self,N1,N2):
        for i in range(1,min(N1,N2)+1):
            if(N1%i==0 and N2%i==0):
                gcd = i
                
        print(gcd)
        
            
if __name__ == "__main__":
    sol = Solution()
    N1 = 9
    N2 = 12
    sol.GCD(N1,N2)
    
```

```python
#Better
import math
class Solution:
    def GCD(self,N1,N2):
        for i in range(min(N1,N2),0,-1):
            if(N1%i==0 and N2%i==0):
                gcd = i
                break
                
        print(gcd)
        
            
if __name__ == "__main__":
    sol = Solution()
    N1 = 9
    N2 = 12
    sol.GCD(N1,N2)
    
```
